```python


"({[]})"
"([)]"
"((()))"
"({)}"

def is_balanced(s):
    while "()" in s or "[]" in s or "{}" in s:
        s = s.replace("()", "").replace("[]", "").replace("{}", "")
    return s == ""

print(is_balanced("({[]})")) # True
print(is_balanced("([)]"))   # False
print(is_balanced("((()))")) # True
print(is_balanced("({)}"))   # False
```
