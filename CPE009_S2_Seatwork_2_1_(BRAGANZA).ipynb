```python
# Overall changes

def term_crit(actNumber):  # better name for readability
    rawScore = []
    totalScore = []

    for i in range(actNumber):  # removed the 0 (it essentially does the same thing and removes clutter)
        inputScore = float(input("Score: "))
        inputTotal = float(input("Total: "))

        if inputScore > inputTotal:
            print("\nInvalid Scores.\n")
            return None  # doesn't compute the invalid input

        rawScore.append(inputScore)
        totalScore.append(inputTotal)

    if sum(totalScore) == 0:
        return 0

    grade = (sum(rawScore) / sum(totalScore)) * 100  # shorter/better readability
    return grade
```

```python
# Overall changes
# Changed some of the naming conventions for better readability
# implemented "return None" basically means “no valid value was returned.” to prevent errors
def term_cs():
    print("Hands-on Activities: ")
    term_hoa = term_crit(int(input("How many hands-on activities this term? ")))
    if term_hoa is None:
        return None

    print("Quizzes: ")
    term_quiz = term_crit(int(input("How many quizzes this term? ")))
    if term_quiz is None:
        return None

    print("Assignments: ")
    term_ass = term_crit(int(input("How many assignments? ")))
    if term_ass is None:
        return None

    grade = (0.5 * term_hoa) + (0.3 * term_quiz) + (0.2 * term_ass)
    return grade
```

```python
# Overall changes
# I just added error checking, better naming conventions and better indentions for better readability
def term_grade():
    print("Computing for the term grade.")

    exam_raw = float(input("Exam Score: "))
    exam_total = float(input("Exam Total: "))

    if exam_total == 0:
        print("Invalid exam total!")
        return None

    exam_grade = (exam_raw / exam_total) * 100

    cs_grade = term_cs()
    if cs_grade is None:
        return None

    grade = (0.5 * exam_grade) + (0.5 * cs_grade)
    return grade
```

```python
# Overall changes
# Same procedure as the last block of code.

def semester_grade():
    print("Prelim Grade")
    prelim_grade = term_grade()
    if prelim_grade is None:
        return None

    print("Midterm Grade")
    midterm_grade = term_grade()
    if midterm_grade is None:
        return None

    final_mg = (1/3) * prelim_grade + (2/3) * midterm_grade

    print("Final Grade")
    final_grade = term_grade()
    if final_grade is None:
        return None

    final_fg = (1/3) * final_mg + (2/3) * final_grade

    return final_fg
```

```python
#Overall changes
# Instead of 1 being the only grade given I have added more options to it (I copied and pasted this from the last activity)
def convert_grade(final_grade):
    if 99 <= final_grade <= 100:
        return 1.0
    elif 95 < final_grade < 99:
        return 1.25
    elif 92 < final_grade < 96:
        return 1.5
    elif 89 < final_grade < 93:
        return 1.75
    elif 86 < final_grade < 90:
        return 2.0
    elif 83 < final_grade < 87:
        return 2.25
    elif 80 < final_grade < 84:
        return 2.5
    elif 77 < final_grade < 81:
        return 2.75
    elif 74 < final_grade < 78:
        return 3.0
    else:
        return 5.0

```

```python
#Overall changes
#I added more error checking for this line (same concept as the others if it returns with an invalid input it will crash)
final_numeric = semester_grade()
if final_numeric is not None:
    print("Grade Point:", convert_grade(final_numeric))
else:
    print("Invalid input detected. Cannot compute grade.")
```
