Develop a python scripts that:

    Calculate total marks and percentage
    Assign grade for each subject based on marks:

90-100: A+

80-89: A

70-79: B+

60-69: B

50-59: C

40-49: D

Below 40: F (Fail)

    Determine overall result (Pass/Fail)



Performance Remarks:


Percentage ≥ 90: "Outstanding"

Percentage ≥ 80: "Excellent"

Percentage ≥ 70: "Very Good"

Percentage ≥ 60: "Good"

Percentage ≥ 40: "Satisfactory"

Below 40: "Needs Improvement"


```python
subject_count = 0 #subject counter
print("-- Welcome to the Grading Script! ---")
print("\n")
while subject_count < 2: #Amount of subjects (I did 2 subjects)
  subject_name = input("Enter Subject: ")
  quiz_grade = float(input("Enter Quiz Mark: "))
  ass_grade = float(input("Enter Asssesment Mark: "))
  exam_grade = float(input("Enter Exam Mark: "))
  subject_grade = quiz_grade*.3 + ass_grade*.2 + exam_grade*.5
  print("Subject Grade: ", subject_grade)
  #Grade System
  if 90 <= subject_grade <= 100:
    print("A+")
    print("Outstanding")
  elif 80 <= subject_grade <= 89:
    print("A")
    print("Excellent")
  elif 70 <= subject_grade <= 79:
    print("B+")
    print("Very Good")
  elif 60 <= subject_grade <= 69:
    print("B")
    print("Very Good")
  elif 50 <= subject_grade <= 59:
    print("C")
    print("Good")
  elif 40 <= subject_grade <= 49:
    print("D")
    print("Satisfactory")
  else:
    print("F")
    print("Needs Improvement")
  print("\n")
  subject_count += 1 #Increments 1 until the < 2 condition is met
```
