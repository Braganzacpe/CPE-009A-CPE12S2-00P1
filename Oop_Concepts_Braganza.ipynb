```python
#Animal's features like name, habitat, and temperature
class Animals:
  def __init__(self,name,habitat,temperature):
    self.name = name
    self.habitat = habitat
    self.temperature = temperature


  def features(self):
    return f"The {self.name} is a {self.temperature}-blooded animal that lives on {self.habitat}."

mammals = Animals("Dog", "Land", "Warm")
birds = Animals("Sparrow", "Air", "Warm")
reptiles = Animals("Snake", "Terrestrial", "Cold")

print(mammals.name)
print(mammals.habitat)

print(birds.name)
print(birds.habitat)

print(reptiles.name)
print(reptiles.habitat)

print(mammals.features())
print(birds.features())
print(reptiles.features())
```

```python
import math

class Shape:
    def __init__(self, name, sides):
        self.name = name.lower()
        self.sides = sides

    def description(self):
        return f"The {self.name.capitalize()} has {self.sides} sides."

    def calculate_area(self, **dimensions):
        formulas = {
            "triangle": lambda d: 0.5 * d.get('base', 0) * d.get('height', 0),
            "circle": lambda d: math.pi * (d.get('radius', 0) ** 2),
            "square": lambda d: d.get('side', 0) ** 2,
            "rectangle": lambda d: d.get('length', 0) * d.get('width', 0)
        }

        if self.name in formulas:
            return formulas[self.name](dimensions)
        else:
            raise ValueError(f"Area formula for {self.name} is not defined.")


# Create objects
shapes = [
    Shape("Triangle", 3),
    Shape("Circle", 1),
    Shape("Square", 4),
    Shape("Rectangle", 4)
]

# NEW dimensions (changed values)
dimensions_list = [
    {"base": 12, "height": 6},
    {"radius": 7},
    {"side": 8},
    {"length": 9, "width": 4}
]

# Loop through shapes
for shape, dims in zip(shapes, dimensions_list):
    print(shape.description())
    area = shape.calculate_area(**dims)

    if shape.name == "circle":
        print(f"Area of {shape.name.capitalize()}: {area:.2f}")
    else:
        print(f"Area of {shape.name.capitalize()}: {area}")
```
